Através do método df.info() foi possível identificar que há 4 colunas inválidas no DataFrame.

```python
# Retirar as colunas que não serão utilizadas na análise
df = df.drop(columns=["Unnamed: 10", "Unnamed: 11", "Unnamed: 12", "Unnamed: 13"])
```

```python
# Exibir os registros onde a coluna "PR_NOME" possui o valor "N/D" para identificar possíveis inconsistências nos dados
display(df[df["PR_NOME"] == "#N/D"])
```

Através do método display() acima, foi possível verificar que 3650 linhas do DataFrame possuem registro com valor "#N/D" nas colunas "PR_CAT" e "PR_NOME"

```python
# Descartar os registros onde a coluna "PR_NOME" possui o valor "N/D" para evitar inconsistências na análise
df[["PR_CAT", "PR_NOME"]] = df[["PR_CAT", "PR_NOME"]].replace("#N/D", np.nan)
df = df.dropna(subset=["PR_CAT", "PR_NOME"])
```
