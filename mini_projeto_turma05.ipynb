```python
# Inserir todos os imports necessários para o projeto
import pandas as pd
import numpy as np
```

```python
# Carregar o arquivo CSV com os dados do projeto
df = pd.read_csv(r"C:\Users\LUIZ HENRIQUE\sesi_senai\mini_projeto_turma05\Base Varejo.csv", sep=";")
df.head() # Exibir as primeiras linhas do DataFrame para verificar se os dados foram carregados corretamente
```

```python
# Exibir informações sobre o DataFrame, como número de linhas, colunas e tipos de dados
# Ponto de atenção: Os valores da coluna DATA estão como string. É necessário converter para o tipo datetime.
df.info()
```

```python
# Converter a coluna 'DATA' para o tipo datetime
df['DATA'] = pd.to_datetime(df['DATA'], format='%d/%m/%Y', errors='coerce')
df.info() # Confirmar que a conversão foi bem-sucedida e que a coluna 'DATA' agora está no formato datetime.
df.head() # Exibir linhas para confirmar que a coluna 'DATA' foi convertida corretamente.
```

Através do método df.info() foi possível identificar que há 4 colunas inválidas no DataFrame.
São elas: Unnamed: 10, Unnamed: 11, Unnamed: 12 e Unnamed: 13.

```python
# Retirar as colunas que não serão utilizadas na análise
df = df.drop(columns=["Unnamed: 10", "Unnamed: 11", "Unnamed: 12", "Unnamed: 13"])
```

```python
# Exibir o número de linhas e colunas do DataFrame após a remoção das colunas inválidas
df.shape
```

```python
# Verificar se há registros ausentes no DataFrame
df.isna().sum()
```

```python
# Verificar se há registros nulos no DataFrame
df.isnull().sum()
```

```python
# Exibir estatísticas descritivas do DataFrame, como média, desvio padrão, valores mínimo e máximo, entre outros
display(df.describe())
```

```python
# Exibir os registros onde a coluna "PR_NOME" possui o valor "N/D" para identificar possíveis inconsistências nos dados
display(df[df["PR_NOME"] == "#N/D"])
```

Através do método display() acima, foi possível verificar que 3650 linhas do DataFrame possuem registro com valor "#N/D" nas colunas "PR_CAT" e "PR_NOME"

```python
# Descartar os registros onde a coluna "PR_NOME" possui o valor "N/D" para evitar inconsistências na análise
df[["PR_CAT", "PR_NOME"]] = df[["PR_CAT", "PR_NOME"]].replace("#N/D", np.nan)
df = df.dropna(subset=["PR_CAT", "PR_NOME"])
```

```python
# Exibir o número de linhas e colunas do DataFrame após a remoção dos registros inválidos
df.shape
```

No arquivo .zip encontramos também um arquivo .pdf no qual podemos verificar o que cada coluna do DataSet significa.
Aplicar uma nova nomenclatura para que fique mais fácil a nossa leitura dos dados:

CO_ID     -> ID_COMPRA  
CL_ID     -> ID_CLIENTE  
CL_GENERO -> GENERO_CLIENTE  
CL_EC     -> ESTADO_CIVIL  
CL_FHL    -> FILHOS_CLIENTE  
CL_SEG    -> CLASSE_ECONOMICA_CLIENTE  
PR_ID     -> ID_PRODUTO  
PR_CAT    -> CATEGORIA_PRODUTO  
PR_NOME   -> NOME_PRODUTO  

```python
# Renomear as colunas conforme a convenção acima
df = df.rename(columns={
    "CO_ID": "ID_COMPRA",
    "CL_ID": "ID_CLIENTE",
    "CL_GENERO": "GENERO_CLIENTE",
    "CL_EC": "ESTADO_CIVIL",
    "CL_FHL": "FILHOS_CLIENTE",
    "CL_SEG": "CLASSE_ECONOMICA_CLIENTE",
    "PR_ID": "ID_PRODUTO",
    "PR_CAT": "CATEGORIA_PRODUTO",
    "PR_NOME": "NOME_PRODUTO",
})
# Verificar as colunas do DataFrame após a renomeação
df.head()
```
